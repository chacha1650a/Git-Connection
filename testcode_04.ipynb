```python
with open("log.txt", "w", encoding="utf-8") as f:
    f.write("경보1: 1.1.1.1\n")   #← 파일에 쓰기
                            # ← with 블록 끝 → 자동 닫힘

with open("log.txt", encoding="utf-8") as f:
    for line in f:                 #← 한 줄씩 읽기
        print(line.strip())        #← 줄바꿈 제거 후 출력
```
